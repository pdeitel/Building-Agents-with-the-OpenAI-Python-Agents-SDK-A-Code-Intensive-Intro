## OpenAI Python Ecosystem Overview
* **OpenAI Python SDK** for direct API calls from Python programs and notebooks
    * https://github.com/openai/openai-python
    * https://developers.openai.com/api/docs/
* **OpenAI Python Agents SDK** for code-first agent workflows with tools, handoffs, state, and observability
    * https://github.com/openai/openai-agents-python
    * https://developers.openai.com/api/docs/guides/agents/quickstart
* **OpenAI Agents API** (public beta, September 10, 2026) — OpenAI runs the agent loop for you in the cloud, using the same Codex harness that powers Codex
    * Managed sessions, context compaction, subagents, tools/MCP and hosted sandboxes for long-running agents
    * Accessed through the **OpenAI Python SDK**, not the Agents SDK: `client.beta.agents.sessions.create(...)`
    * https://developers.openai.com/api/docs/guides/agents-api/overview
* **Codex** as a coding agent for working on real software projects and course code
    * Also, enables capabilities like controlling your local computer
    * https://learn.chatgpt.com/docs
---

## OpenAI Python SDK vs. OpenAI Agents SDK
* **OpenAI Python SDK** — for calling OpenAI APIs directly from your Python apps
    * **Responses API** is the central API: prompt a model, process its returned content
    * Text and code generation, vision and multimodal prompts, structured outputs, function calling, audio, images, moderation
    * **Your app owns the control flow** — it decides when to call tools and what happens next
* **OpenAI Agents SDK** — for apps that need more orchestration than a single API call
    * Agents plan multi-step control flow, call tools, hand off to other agents, and manage conversation state
    * Work autonomously or with a human in the loop
    * **The SDK owns the agent loop** — and uses the Responses API under the hood

## OpenAI Agents API 
* Public beta: Sept 10, 2026
* **Codex/Work harness as an OpenAI-managed service**
    * Exposed by API calls
    * OpenAI-hosted sandbox runs the agent loop: Session orchestration, context compaction, recovery from failures
    * 9 partner sandbox providers
        * Blaxel, Cloudflare, Daytona, DigitalOcean, E2B, Modal, Oracle, Runloop, Vercel
* **Pricing:** no separate fee — pay for model tokens, OpenAI tools and sandbox container time
* **Beta caveats:** US data residency only; no Zero Data Retention yet
* https://developers.openai.com/api/docs/guides/agents-api/overview

## Models
* [View All Models](https://developers.openai.com/api/docs/models/all)
* **GPT-5.6 replaced the old full/`mini`/`nano` naming with three named tiers**
    * `gpt-5.6-sol` — the GPT-5.6 flagship for complex professional work; now sits below Astra
    * `gpt-5.6-terra` — balances intelligence and cost; the everyday default
    * `gpt-5.6-luna` — fastest and cheapest, for high-volume workloads
    * The bare alias `gpt-5.6` routes to `gpt-5.6-sol`
* **All 5.6 tiers reason**
    * Unlike `nano` models, Luna supports reasoning effort from `none` through `max`, and all three share a 1,050,000-token context window
    * Set `reasoning={'effort': 'none'}` when you want old `nano`-style speed and cost
* **GPT-6 Astra** (`gpt-6-astra`, released September 3, 2026) 
    * Most capable model 
    * Long, complex, multi-step work: coding, browser and computer use, analysis, science
    * Only GPT-6 model so far
    * More expensive than 5.6 models
    * https://developers.openai.com/api/docs/models/gpt-6-astra
