# Creating Agents with the OpenAI Agents SDK
* OpenAI Agents SDK is a lightweight package for building agentic apps
* Orchestrate multi-step work 
 
---

## Agents SDK Features
* **Agent** — instructions + model + tools; stateless by default
* **Tools** — hosted tools, Python function tools, MCP servers, other agents
* **State** — lets multi-turn work preserve useful context
* **Handoffs** — transfer control to other agents
* **Guardrails** — input/output validation and safety checks
* **Tracing** — logs every step for debugging and evaluation
* **Runner** — runs the **agent loop** (next slides)

---

## Beyond Basic Agents
* **Sandbox Agents** (beta) — files, shell, memory and **Skills** in a controlled workspace
    * https://openai.github.io/openai-agents-python/sandbox_agents/
* **Realtime Agents** — the SDK's voice agents: `RealtimeAgent` + `RealtimeRunner` on the Realtime API
    * One model (`gpt-realtime-2.1`) handles speech, reasoning *and* tool calls
    * https://openai.github.io/openai-agents-python/realtime/quickstart/
* **GPT-Live** — full-duplex voice model that **delegates** reasoning and tools to a backend agent
    * Not an SDK feature: reached via `client.live` in the OpenAI Python SDK; demoed in `02-05-10`

---

## Default Model
* **`gpt-5.6-luna`** is the SDK's built-in default — no reasoning, low verbosity, for cheap, high-volume work
* This course's examples set `model='gpt-6.1-terra'`
* Changing the model
    * `OPENAI_DEFAULT_MODEL` environment variable — new default for agents that don't set a model ([docs](https://openai.github.io/openai-agents-python/models/#default-model))
    * `Runner.run(..., run_config=RunConfig(model=...))` — overrides every agent's model for one run ([docs](https://openai.github.io/openai-agents-python/running_agents/#run-config))

---

## The Agent Loop: ReAct — **Re**asoning + **Act**ing
* Model reasons, calls a tool, observes the result, then returns an answer or updates its plan and continues
* Example — "How many Apple shares can I buy with €1,000?" (two trips through the loop)
    * Reason → "I need AAPL's current price."
    * Act → call tool `get_quote('AAPL')`
    * Observe → `{'last_price': 211.34, 'currency': 'USD', ...}`
    * Reason → "The price is in US dollars but the budget is in euros, so I need the exchange rate."
    * Act → call tool `get_exchange_rate('EUR', 'USD')`
    * Observe → `{'rate': 1.08}`
    * Reason → "€1,000 is \\$1080, and \\$1080 ÷ \\$211.34 ≈ 5.1 shares."
    * Answer → "You can buy 5 whole shares of Apple with €1,000."
* Each observation feeds the next reasoning step
* If a tool fails, the model can try a different approach — no special error-handling code
* OpenAI doesn't use the term ReAct, but its agent loop works this way
* Sources: [Yao et al., 2022](https://arxiv.org/abs/2210.03629) · [Agents SDK: Running agents](https://openai.github.io/openai-agents-python/running_agents/) · [IBM: What is a ReAct agent?](https://www.ibm.com/think/topics/react-agent)
